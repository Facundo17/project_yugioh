# Este notebook divide prepara las imágenes

Las divide en tres carpetas por tipo de carta:

Monster, Spell y Trap

```python
import os
import shutil
import pandas as pd

# Configuración
DOWNLOAD_DIR = "yugioh_dataset"
OUTPUT_DIR = "yugioh_dataset_cleaned"
LABEL_MAP = {
    "Monster": "Monster",
    "Spell": "Spell",
    "Trap": "Trap"
}

# Leer el CSV con información de las cartas
csv_path = os.path.join(DOWNLOAD_DIR, "Yugi_db_cleaned.csv")
df = pd.read_csv(csv_path)

df.columns.to_list()

#Crear carpetas de salida
os.makedirs(OUTPUT_DIR, exist_ok=True)
for label in LABEL_MAP.values():
    os.makedirs(os.path.join(OUTPUT_DIR, label), exist_ok=True)

# Clasificar imágenes por tipo
image_dir = os.path.join(DOWNLOAD_DIR, "Yugi_images")
errors = []

print("Clasificando imágenes...")

for _, row in df.iterrows():
    card_type = row.get("Card type", "")
    image_name = row.get("Image_name", "")
    
    # Filtrar por tipo válido
    if not isinstance(card_type, str) or not isinstance(image_name, str):
        continue

    if "Monster" in card_type:
        label = "Monster"
    elif "Spell" in card_type:
        label = "Spell"
    elif "Trap" in card_type:
        label = "Trap"
    else:
        continue  # tipo no reconocido

    src_path = os.path.join(image_dir, image_name)
    dst_path = os.path.join(OUTPUT_DIR, label, image_name)

    if os.path.exists(src_path):
        shutil.copy2(src_path, dst_path)
    else:
        errors.append(image_name)

print(f"✅ Clasificación terminada. Errores: {len(errors)}")
```
